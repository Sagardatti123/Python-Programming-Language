```python
def fun(a):
  vowels='AEIOUaeiou'
  result=''
  for ch in a:
    if ch not in vowels:
      result=result+ch
  return result
res=fun('hai')
print(res)


```

```python
def sagar(string):
  vowels='AEIOUaeiou'
  res=''
  for ch in string:
    if ch in vowels:
      res=res+ch
  return f'vowels of string:{res}'
result=sagar("data")
print(result)


```

```python
def sagar(num):

  result=[]
  for i in num:
    a=i*10+num.index(i)
    result.append(a)
  return result

res=sagar([10,11,50,34])
print(res)
```

```python
def div():

  res=[]
  for i in range(1,50):
    if i%2!=0 and i%3!=0:
      res.append(i)
  return res

result=div()
print(result)
```

```python
def word(string):
  words=string.split()
  result=[]
  for word in words:
    if len(word)==1:
      pass
    else:
      new=word[0].upper()+word[1:-1]+word[-1].upper()
      result.append(new)
  return ' '.join(result)
text=input("enter s string:") #sagar datti
print(word(text))
```
