```python
def name(string):
  dict={}
  for ch in string:
    if ch not in dict:
      dict[ch]=1
    else:
      dict[ch]=dict[ch]+1
  return dict
text=input("enter a text:")#sagar datti
print(name(text))

```

```python
import math
def area(a,b,c):
  s=a+b+c/2
  res=math.sqrt(s*(s-a)*(s-b)*(s-c))
  return res
result=area(2,3,4)
```

```python
def avg(num):
  res=0
  count=len(str(num))

  while num >0:
    a=num%10
    res=res+a

    num=num//10
  result=res/count
  return result
n=int(input("enter a value:"))
print(avg(n))
```

```python
def sum(n):
  res=0
  while n>0:
    a=n%10
    res=res+a
    n=n//10
  return f'sum of digits is :{res}'

n=int(input('enter a number:'))
print(sum(n))
```

```python
def prime(n):
  count=0
  for i in range(1,n+1):
    if n%i==0:
      count=count+1
  if count==2:
    print("prime")
  else:
    print("not prime")
prime(7)
```

```python
def fact_number(n):
  fact=1
  for i in range(1,n+1):
    fact=fact*i
  return  f'factorial of a number is :{fact}'
num=int(input('enter number:'))
print(fact_number(num))
```

```python
def palindrome(string):
  rev=''
  for ch in string:
    rev=ch+rev
  if rev==string:
    print("palimdrome")
  else:
    print('not a palindrome')
text=input("enter a string:")
palindrome(text)
```

```python
def upper_case(string):
  count=0
  for ch in string[:4]:
    if ch.isupper():
      count+=1
  if count>=2:
    return string.upper()
  else:
    return string
text=input()
print(upper_case(text))
```

```python
def spaces(string):
  result=''
  spaces=0
  for ch in string:
    if ch==' ':
      spaces=spaces+1
    else:
      result=result+ch

  return ' '*spaces+result
text=input('enter a text:')
print(spaces(text))




```

```python
def sagar(string):
  upper_case=0
  lower_case=0
  numbers=0
  space=0
  for ch in string:
    if ch.isupper():
      upper_case+=1
    elif ch.islower():
      lower_case+=1
    elif ch.isnumeric():
      numbers+=1
    else:
      space+=1
  print(f'uppercase:{ upper_case}')
  print(f'lowercase:{ lower_case}')
  print(f'numbers:{numbers}')
  print(f'spaces:{space}')
sagar('data sCience 12444')

```
