```python
def sagar(string):
  vowels='AEIOUaeiou'
  count=0
  for ch in string:
    if ch in vowels:
      count+=1
  return count
string=input('enter a string:')
print(sagar(string))
```

```python
def datti(a):
  dict={}
  for ch in range(len(a)):
    dict[ch]=a[ch]
  print(dict)
datti('kamal')

```
